```python
class BankAccount:
  AccStatus = "YES"
  def __init__(self,owner,accno,balance):
    self.owner = owner
    self.accno = accno
    self.balance = balance

  def deposit(self,amount):
    if amount>=0:
      self.balance += amount
      print(f"The amount deposited is {amount:.2f} and your balance is {self.balance:.2f}")
    else:
      print("The entered amount is negative")

  def withdraw(self,amount):
    if amount>=0:
      if self.balance >= amount:
        self.balance-=amount
        print(f"You have withdrawn {amount}")
        print(f"Account holder: {self.owner}({self.accno}) and your balance is {self.balance}")
      else:
        print(f"Your balance is not sufficient. Balance:{self.balance:.2f}")
    else:
      print("The amount is negative")

  def display(self):
    print(f"Owner:{self.owner}, Account No: {self.accno}, Balance: {self.balance}")

account1 = BankAccount("Yatheesh",1500, 10000)
account1.display()

account1.withdraw(500)
```
